```python
import os
from google import genai
from google.colab import userdata

# -------------------------------------------------------------------
# SETUP: Fetching the key safely from Colab Secrets
# -------------------------------------------------------------------
# This grabs the secret you just configured in the left panel
if "GEMINI_API_KEY" in dir(userdata):
    os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")

API_KEY = os.environ.get("GEMINI_API_KEY")

# Provide the api_key to the client
client = genai.Client(api_key=API_KEY)

def generate_business_names(niche, keywords, count=5):
    """
    Generates business name ideas using Google's Gemini.
    """
    print(f"Generating ideas for a {niche} business...")

    prompt = (
        f"Generate {count} creative, catchy, and professional business names "
        f"for a {niche} company. Consider incorporating these themes or keywords: {keywords}. "
        "Return the response as a numbered list."
    )
    try:
        # Make the API call to generate text
        response = client.models.generate_content(
            model='gemini-2.5-flash',
            contents=prompt,
        )

        # Print the results
        print("\nHere are your business name ideas:\n")
        print(response.text)

    except Exception as e:
        print(f"\nAn error occurred: {e}")

if __name__ == "__main__":
    print("Welcome to the AI Business Name Generator (Gemini Version)!\n")

    # --- Try it out! You can change these to whatever you want ---
    my_business_niche = "tech startup"
    my_keywords = "AI, fast, future, smart"

    generate_business_names(my_business_niche, my_keywords)
```
